```python

# Prepare the dataset for analysis

# Keep the original dataset unchanged
df_clean = df.copy()

# Give columns shorter, easier-to-use names.
# This follows the 11-column order shown in your screenshot.
df_clean.columns = [
    "customer_id",
    "year",
    "quarter",
    "survey_date",
    "response_date",
    "recommend_score",
    "value_rating",
    "usage_frequency",
    "product_quality",
    "panel_usability",
    "reporting_understanding"
]

# Convert date columns to datetime format
df_clean["survey_date"] = pd.to_datetime(
    df_clean["survey_date"], errors="coerce"
)

df_clean["response_date"] = pd.to_datetime(
    df_clean["response_date"], errors="coerce"
)

# Convert survey ratings into numeric values
rating_columns = [
    "recommend_score",
    "value_rating",
    "product_quality",
    "panel_usability"
]

for col in rating_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col], errors="coerce"
    )

# Display the updated dataset
print("Dataset shape:", df_clean.shape)
print("\nUpdated column names:")
print(df_clean.columns.tolist())

display(df_clean.head())
```

```python

# Data quality assessment

print("Missing values in each column:")
display(df_clean.isnull().sum().to_frame("Missing Count"))

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nRating summaries:")
display(df_clean[rating_columns].describe())

print("\nUnique usage frequencies:")
display(df_clean["usage_frequency"].value_counts(dropna=False))

print("\nSurvey years:")
display(df_clean["year"].value_counts(dropna=False).sort_index())
```

```python

# Check recommendation score values

print("Recommendation score counts:")
display(
    df_clean["recommend_score"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nMinimum recommendation score:",
      df_clean["recommend_score"].min())

print("Maximum recommendation score:",
      df_clean["recommend_score"].max())
```

```python

# Visualize recommendation scores

score_counts = (
    df_clean["recommend_score"]
    .dropna()
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=score_counts.index,
    y=score_counts.values
)

plt.title("Distribution of Customer Recommendation Scores")
plt.xlabel("Recommendation Score")
plt.ylabel("Number of Responses")
plt.tight_layout()
plt.show()
```

```python

# Summary of customer ratings

rating_columns = [
    "recommend_score",
    "value_rating",
    "product_quality",
    "panel_usability"
]

rating_summary = df_clean[rating_columns].agg(
    ["count", "mean", "median", "std", "min", "max"]
).T

rating_summary = rating_summary.round(2)

display(rating_summary)
```

```python

# Compare average scores

average_ratings = df_clean[rating_columns].mean().sort_values()

plt.figure(figsize=(10, 6))

sns.barplot(
    x=average_ratings.values,
    y=average_ratings.index
)

plt.title("Average Scores Across Customer Survey Measures")
plt.xlabel("Average Score")
plt.ylabel("Survey Measure")
plt.tight_layout()
plt.show()
```

```python

# Average recommendation score by year and quarter

quarterly_scores = (
    df_clean
    .groupby(["year", "quarter"])["recommend_score"]
    .agg(["mean", "count"])
    .reset_index()
)

quarterly_scores.columns = [
    "Year", "Quarter", "Average Recommendation Score", "Responses"
]

quarterly_scores["Average Recommendation Score"] = (
    quarterly_scores["Average Recommendation Score"].round(2)
)

display(quarterly_scores)
```

```python

# Visualize quarterly recommendation scores

quarterly_scores["Period"] = (
    quarterly_scores["Year"].astype(str)
    + " Q"
    + quarterly_scores["Quarter"].astype(str)
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=quarterly_scores,
    x="Period",
    y="Average Recommendation Score",
    marker="o"
)

plt.title("Customer Recommendation Scores Across Quarters")
plt.xlabel("Year and Quarter")
plt.ylabel("Average Recommendation Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Compare product quality and panel usability

comparison = df_clean[
    ["product_quality", "panel_usability"]
].agg(["count", "mean", "median", "std"]).T.round(2)

display(comparison)
```

```python

# Distribution of quality and usability ratings

rating_data = df_clean[
    ["product_quality", "panel_usability"]
].melt(
    var_name="Survey Measure",
    value_name="Rating"
)

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=rating_data,
    x="Survey Measure",
    y="Rating"
)

plt.title("Distribution of Product Quality and Panel Usability Ratings")
plt.xlabel("Survey Measure")
plt.ylabel("Rating")
plt.tight_layout()
plt.show()
```

```python

# Distribution of usage frequency

usage_counts = df_clean["usage_frequency"].value_counts(
    dropna=False
)

display(usage_counts.to_frame("Number of Responses"))

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df_clean,
    x="usage_frequency",
    order=df_clean["usage_frequency"].value_counts().index
)

plt.title("Distribution of Customer Usage Frequency")
plt.xlabel("Usage Frequency")
plt.ylabel("Number of Responses")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Recommendation scores by usage frequency

usage_analysis = (
    df_clean.groupby("usage_frequency", dropna=False)
    ["recommend_score"]
    .agg(["count", "mean", "median"])
    .round(2)
    .sort_values("mean", ascending=False)
)

display(usage_analysis)
```

```python

# Correlation analysis

numeric_columns = [
    "recommend_score",
    "value_rating",
    "product_quality",
    "panel_usability"
]

correlation_matrix = df_clean[numeric_columns].corr()

display(correlation_matrix.round(2))

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    fmt=".2f"
)

plt.title("Correlation Between Customer Survey Measures")
plt.tight_layout()
plt.show()
```

## Methodology

The project follows an exploratory data analysis approach to examine customer feedback and satisfaction.

1. **Data loading:** The CSV dataset was imported into Python using Pandas.
2. **Data inspection:** The dataset dimensions, column names, data types, sample records, and unique values were examined.
3. **Data preparation:** Relevant columns were prepared for analysis, including date and numerical conversions.
4. **Data quality assessment:** Missing values and duplicate records were examined.
5. **Descriptive analysis:** Summary statistics were calculated for the available customer survey measures.
6. **Data visualization:** Bar charts, line charts, box plots, and a correlation heatmap were used to explore distributions, changes over time, and relationships between variables.
7. **Interpretation:** The results were used to identify patterns and potential areas for improvement.

## Key Findings and Discussion

The exploratory data analysis was conducted on 1,989 survey records containing 11 variables.

The mean customer recommendation score was 8.30 out of 10, suggesting a generally positive willingness to recommend the platform. The average perceived value rating was 4.27, product quality was 4.12, and panel usability was 4.04. Panel usability had 1,925 non-missing responses.

Correlation analysis identified a positive relationship between recommendation score and perceived value (r = 0.75). Product quality also showed a positive correlation with recommendation score (r = 0.63), followed by panel usability (r = 0.53). These results indicate that respondents who gave higher ratings in these areas also tended to give higher recommendation scores. However, the relationships do not establish causation.

Usage-frequency analysis showed that daily users formed the largest group, with 939 responses, followed by weekly users with 770 and monthly users with 277. Three responses had missing usage-frequency values.

The quarterly average recommendation score declined from 8.41 in Q1 2022 to 8.32 in Q2 and 8.20 in Q3. This modest decline suggests that customer recommendation scores should continue to be monitored over time.

The initial data-quality assessment identified 67 missing cells and seven exact duplicate rows. Further investigation is needed to determine the appropriate treatment of these records.
